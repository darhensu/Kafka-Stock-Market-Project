```python
import pandas as pd
from kafka import KafkaProducer
from time import sleep
import time
from json import dumps
import json
```

```python
# Konfigurasi produser Kafka
producer = KafkaProducer(
    bootstrap_servers=['18.207.190.183:9092'],
    value_serializer=lambda x: dumps(x).encode('utf-8')
)
```

```python
# Fungsi untuk mengirim pesan
def send_message(topic, value):
    producer.send(
        topic=topic,
        value=value  # value sudah dalam format JSON yang diserialisasi
    )
    producer.flush()  # Pastikan pesan terkirim sebelum melanjutkan
```

```python
# Kirim pesan ke topik 'demo_test'
send_message('demo_test', {'halo teman2 status whatsappku yang gabut melihat statusku':''})

print("Pesan berhasil dikirim ke Kafka!")
```

```python
# Function to send messages
def send_messages():
    while True:
        try:
            # Sample a random record from the DataFrame
            dict_stock = df.sample(1).to_dict(orient="records")[0]

            # Send the record to Kafka topic 'demo_test'
            producer.send('demo_test', value=dict_stock)
            print(f"Sent message: {dict_stock}")

            # Flush to ensure the message is sent
            producer.flush()

            # Add a short delay (adjust as needed)
            time.sleep(1)

        except Exception as e:
            print(f"Error sending message: {e}")
```

```python
# Start sending messages
send_messages()
```
