```python
import pandas as pd
from kafka import KafkaConsumer
from time import sleep
from json import dumps, loads, JSONDecodeError
import json
from s3fs import S3FileSystem
```

```python
# Konfigurasi consumer Kafka
consumer = KafkaConsumer(
    'demo_test',  # Nama topik yang ingin didengarkan
    bootstrap_servers=['18.207.190.183:9092'],  # Alamat server Kafka
    auto_offset_reset='earliest',  # Membaca pesan dari awal
    value_deserializer=lambda x: x.decode('utf-8')  # Hanya mendekode menjadi string UTF-8
)
```

```python
# Fungsi untuk membaca pesan dari Kafka dengan error handling
def consume_message():
    for message in consumer:
        try:
            # Mencoba untuk mengubah pesan menjadi JSON
            value = loads(message.value)
            print(f"Pesan diterima: {value}")
        except JSONDecodeError:
            # Menangani kesalahan saat deserialisasi JSON
            print(f"Pesan tidak valid: {message.value}")
```
